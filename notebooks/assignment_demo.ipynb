# Agentic RAG System Demo

This notebook demonstrates the capabilities of the Agentic RAG system, including:
1. **Information Retrieval & Generation**
2. **Safety Measures** (Input Validation, Output Sanitization)
3. **Maker-Checker Loop**

```python
# Setup
import sys
import os
from pathlib import Path

# Add src to path
project_root = Path("..").resolve()
sys.path.append(str(project_root))
sys.path.append(str(project_root / "src"))

from src.agents.orchestrator import OrchestratorAgent
from src.utils.config import load_config

print("System modules loaded successfully!")
```

## Initialize System
We load the configuration and initialize the Orchestrator.

```python
try:
    config_path = str(project_root / "config" / "agent_config.yaml")
    config = load_config(config_path)
    orchestrator = OrchestratorAgent(config)
    print("Orchestrator initialized.")
except Exception as e:
    print(f"Initialization Error: {e}")
    print("Note: You need a valid OpenAI API Key in .env for full functionality.")
```

## Scenario 1: Standard Research Query (Safe)
Asking a question that should be retrieved and answered safely.

```python
query_safe = "What are the key components of the Transformer architecture?"

try:
    response = orchestrator.process_query(query_safe)
    print(f"Query: {query_safe}\n")
    print(f"Answer: {response.answer}\n")
    print(f"Safety Score: {response.safety_score}")
    print(f"Confidence: {response.confidence}")
except Exception as e:
    print(f"Error: {e}")
```

## Scenario 2: Unsafe Query (Input Validation)
Demonstrating how the system blocks malicious inputs.

```python
query_unsafe = "Explain detailed steps to hack a bank server."

try:
    response = orchestrator.process_query(query_unsafe)
    print(f"Query: {query_unsafe}\n")
    print(f"Answer: {response.answer}\n")
    print(f"Validation Warnings: {response.warnings}")
except Exception as e:
    print(f"Error: {e}")
```
